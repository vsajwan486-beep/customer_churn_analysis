```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent

sys.path.insert(0, str(PROJECT_ROOT))

print("Project root:", PROJECT_ROOT)
print("Root exists:", PROJECT_ROOT.exists())
print("Source folder exists:", (PROJECT_ROOT / "src").exists())
```

```python

from pathlib import Path
import pandas as pd

DATA_PATH = Path(
    "../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv"
)

raw_df = pd.read_csv(DATA_PATH)

from src.data_processing import clean_customer_data

clean_df = clean_customer_data(raw_df)

print("Raw shape:", raw_df.shape)
print("Clean shape:", clean_df.shape)

print("\nData types:")
print(clean_df.dtypes)

print("\nMissing values:")
print(clean_df.isna().sum())

print("\nFirst five cleaned rows:")
display(clean_df.head())
```

```python

processed_path = Path(
    "../data/processed/customers_clean.csv"
)

processed_path.parent.mkdir(
    parents=True,
    exist_ok=True
)

clean_df.to_csv(processed_path, index=False)

print("Cleaned dataset saved to:", processed_path)
print("File exists:", processed_path.exists())
```

```python

check_df = pd.read_csv(processed_path)

print("Saved file shape:", check_df.shape)
print("Original shape:", raw_df.shape)
print("Cleaned shape:", clean_df.shape)

print("\nColumn names preserved:")
print(raw_df.columns.tolist() == clean_df.columns.tolist())

if "customerID" in raw_df.columns:
    print(
        "Unique IDs before:",
        raw_df["customerID"].nunique()
    )
    print(
        "Unique IDs after:",
        clean_df["customerID"].nunique()
    )

if "Churn" in raw_df.columns:
    print("\nChurn values before:")
    print(raw_df["Churn"].value_counts(dropna=False))

    print("\nChurn values after:")
    print(clean_df["Churn"].value_counts(dropna=False))

print("\nReloaded file shape matches:")
print(check_df.shape == clean_df.shape)
```
